```python
my_list = [1, 2, 3, 4, 5]
# Multiplying each item by 3
result_list = []
for item in my_list :
    result_list . append ( item * 3)



# Python (manual matrix multiplication)
def matrix_multiply (a, b):
    result = []
    for i in range ( len (a)):
        row = []
        for j in range ( len (b [0]) ):
            total = 0
            for k in range ( len (b)):
                total += a[i][k] * b[k][j]
            row.append ( total )
        result.append (row )
    return result

# Example matrices: (2x3) * (3x2) -> (2x2)
A = [
    [1, 2, 3],
    [4, 5, 6],
]
B = [
    [5,  8],
    [9, 10],
    [11, 13],
]

C = matrix_multiply(A, B)

# Print the outcome
print("Product A x B:")
for r in C:
    print(r)
```

```python
#Creating arrays
arr1 = np.array([1, 2, 3, 4, 5]) #1-D arrays
arr2 = np.array([[1, 2, 3], [4, 5, 6]]) #2-D arrays (like tables)
arr3 = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]]) #3-D Arrays
arr_n = np.array([1, 2, 3, 4], ndmin=9) #high-dimensional array
```

```python
# Calculate and print EOQ
D = 1200
S = 100
H = 2

EOQ = np.sqrt((2*D*S)/2)
print(f" EOQ is {EOQ.round(2)}")
```

```python
# Calculate and print Total Cost
total_cost = (D/EOQ)*(S)+(EOQ/2)*(H)
print(f"Total Cost is $ {total_cost.round(2)}")
```
